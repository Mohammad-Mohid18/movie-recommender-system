## Handling Missing Values:

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Not Rated')
```

```python
df = df.dropna(subset=['date_added'])
```

## Seperating Year,Month,Date from Date Column

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month_name()
df['day_added'] = df['date_added'].dt.day
```

```python
text_cols = ['type','title','director','cast','country','rating','listed_in']

for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
df[['duration_value','duration_unit']] = df['duration'].str.split(' ', n=1, expand=True)

df['duration_value'] = pd.to_numeric(df['duration_value'], errors='coerce')
```

```python
df['country'] = df['country'].str.split(',').str[0].str.strip()
```
