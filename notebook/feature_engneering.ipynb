```python
import pandas as pd

df = pd.read_csv("../data/netflix_cleaned.csv")
```

### Select important columns for recomendation:

```python
cols = ['listed_in','cast','director','description','type']

for col in cols:
    df[col] = df[col].str.lower()
```

```python
for col in cols:
    df[col] = df[col].str.replace(',', ' ')
```

### Store the info in a tag for easy tokenization

```python
df['tags'] = (
    df['listed_in'] + ' ' +
    df["listed_in"] + " " +   # weight genre
    df['cast'] + ' ' +
    df['director'] + ' ' +
    df['description'] + ' ' +
    df['type']
)
```

```python
df['tags'] = df['tags'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

```python
df.to_csv("../data/netflix_features.csv", index=False)
```
